```python
import pandas as pd
import numpy as np

## Set up to print every columns and rows
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', None)
```

#### Input txt file
Two files are made by using playground.py and KITplayground.py file in analysis/ directory.   
First line contains like this:

    evtID    ele_pt    ele_phi ...... pho_phi    ID


Contents are like this:

    12140472    77.13    2.92  ...... 232.0    2.500    4
    

```python
# txt files includes numbers from ntuples
#  evtID ele_pt ele_phi ...
file1 = 'sort_KIT_checkID.txt'
file2 = 'sort_COFFEA_checkID.txt'
f1 = 'KIT'
f2 = 'KNU'
```

```python
## Read two files and save df1, df2
df1 = pd.read_csv(file1, sep='\s+')
df2 = pd.read_csv(file2, sep='\s+')
```

```python
## Merge two data frame on event ID 'evtID'
merged_df = pd.merge(df1, df2, on='evtID', suffixes=('_KIT','_KNU'))
merged_df
```

```python
## Compare Electron
column_name = 'ele_pt'
column_name2 = 'ele_eta'
column_name3 = 'ele_phi'
column_name4 = 'ele_ID'
different_values = merged_df[merged_df[f'{column_name}_{f1}'] != merged_df[f'{column_name}_{f2}']]
condition_e = different_values[[f'{column_name}_{f1}', f'{column_name}_{f2}']].isna().all(axis=1)
df_ele = different_values[~condition_e]
df_ele[['evtID', f'{column_name}_{f1}', f'{column_name}_{f2}',f'{column_name2}_{f1}',f'{column_name2}_{f2}',f'{column_name3}_{f1}',f'{column_name3}_{f2}',f'{column_name4}']]
```

```python
## Compare Photon
column_name = 'pho_pt'
column_name2 = 'pho_eta'
column_name3 = 'pho_phi'
column_name4 = 'pho_ID'
different_values = merged_df[merged_df[f'{column_name}_{f1}'] != merged_df[f'{column_name}_{f2}']]
condition_p = different_values[[f'{column_name}_{f1}', f'{column_name}_{f2}']].isna().all(axis=1)
df_pho = different_values[~condition_p]
df_pho[['evtID', f'{column_name}_{f1}', f'{column_name}_{f2}',f'{column_name2}_{f1}',f'{column_name2}_{f2}',f'{column_name3}_{f1}',f'{column_name3}_{f2}',f'{column_name4}']]
```

```python
## Compare Muon
column_name = 'mu_pt'
column_phi = 'mu_phi'
different_values = merged_df[merged_df[f'{column_name}_{f1}'] != merged_df[f'{column_name}_{f2}']]
condition_p = different_values[[f'{column_name}_{f1}', f'{column_name}_{f2}',f'{column_name}_{f1}',f'{column_name}_{f2}']].isna().all(axis=1)
df_pho = different_values[~condition_p]
df_pho[['evtID', f'{column_name}_{f1}', f'{column_name}_{f2}',f'{column_phi}_{f1}',f'{column_phi}_{f2}']]
```

```python
column_name = 'recoil'
def round_to_3_sig_figs(x):
    return round(x, -int(np.floor(np.log10(abs(x)))) + 2)
merged_df[f'recoil_{f1}'] = merged_df[f'recoil_{f1}'].apply(round_to_3_sig_figs)
diff_recoil = merged_df[ abs(merged_df[f'{column_name}_{f1}'] - merged_df[f'{column_name}_{f2}']) > 1]
diff_recoil[['evtID', f'{column_name}_{f1}', f'{column_name}_{f2}']]
```
